# Names and help
> Inspect, list and erase names, and read help

## Names

```bpl
]help •nc
```

```bpl
v←1 2 3 ◇ mean←+/÷≢
•nc "v"
•nc "mean" "absent"
```

```bpl
]help •nl
```

Shadowed names appear once. Implicit argument and operand names are left out:

```bpl
mean←+/÷≢
"me" •nl 3
"absent" •nl 2 3 4
```

```bpl
]help •src
```

For a primitive or derived function, the BPL representation is returned:

```bpl
f←{≥ω} ◇ add←+
•src "f"
•src "add"
```

```bpl
]help •ex
```

A function that another name or an array still holds stays usable after its name is erased. `1` is also returned when the name is already absent:

```bpl
v←1
•ex "v"
•ex "v"
•ex "$a"
```

## Interactive help

`]help name` shows help in the REPL and in notebooks. A leading `⍝` comment block inside a definition supplies its help text:

```bpl
mean←{⍝ Mean of a vector
    +/ω ÷ ≢ω}
```

`]help mean` shows the comment:

```bpl
]help mean
```

Pass `-source` to show the definition:

```bpl
]help mean -source
```

`]help` also accepts a glyph and shows the help from its glyph page:

```bpl
]help ?
```

In the BPL kernel, press Shift-Tab to inspect the name or glyph at the cursor. Press Tab to complete user names, system names and glyph names typed after a backtick, such as `` `iota ``.

`]help` and Shift-Tab read definitions without executing them. For a dfn, the help shows both call forms, `f Y` and `X f Y`, whether or not the dfn takes a left argument.
