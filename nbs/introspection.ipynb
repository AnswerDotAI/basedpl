# Names and help
> Inspect, list and erase names, and read help

## Names

```bpl
]help •nc
```

```bpl
v←1 2 3 ⋄ mean←{+/⍵ ÷ ≢⍵}
•nc "v"
•nc "mean" "absent"
```

```bpl
]help •nl
```

Shadowed names appear once. Implicit argument and operand names are left out:

```bpl
mean←{+/⍵ ÷ ≢⍵}
"me" •nl 3
"absent" •nl 2 3 4
```

```bpl
]help •src
```

Primitives and derived functions give their BPL representation:

```bpl
f←{⍵+1} ⋄ add←+
•src "f"
•src "add"
```

```bpl
]help •ex
```

An already absent name also gives `1`. Retained function handles stay usable:

```bpl
v←1
•ex "v"
•ex "v"
•ex "$a"
```

## Interactive help

`]help name` shows help in the REPL and in notebooks. A leading `⍝` comment block inside a definition supplies its help text:

```bpl
mean←{⍝ Mean of a vector
    +/⍵ ÷ ≢⍵}
```

`]help mean` shows the comment:

```bpl
]help mean
```

`-source` shows the definition:

```bpl
]help mean -source
```

Glyphs work too. Their help comes from the glyph pages:

```bpl
]help ?
```

In the BPL kernel, Shift-Tab inspects the name or glyph at the cursor. Tab completes user names, system names and backtick glyph names.

Inspection reads definitions without executing them. Arbitrary dfns show the one-/two-argument calling convention without inferring their supported valence.
