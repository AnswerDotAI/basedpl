# Superscripts and subscripts

> Marks after an item that apply an operation to it or select from it

A superscript, such as `²`, is a mark written straight after an item that applies an operation to it. A subscript, such as `₁`, selects part of the item instead. Both bind more tightly than any function. `v₁²` is the square of item 1 of `v`:

```bpl
v←10 20 30
v₁²
```

[Typing glyphs](glyphs.qmd#typing-glyphs) gives their keys.

## Superscripts

### On an array

On an array, a superscript digit is a power:

```bpl
x←3
x²+1
```

A power of `¯1`, written `⁻¹`, is the reciprocal:

```bpl
2⁻¹
```

`ᵀ` swaps rows and columns:

```bpl
m←[1 2
   3 4]
mᵀ
```

In `2 3²`, the superscript touches only the `3`:

```bpl
2 3²
```

In `[2 3]²`, it touches the whole list:

```bpl
[2 3]²
```

After a space, a superscript applies to the whole run on its left:

```bpl
2 3 ²
```

The run can be any expression, such as `x+1`:

```bpl
x+1 ²
```

A strand of names is several runs, one for each name:

```bpl
y←5
x y ²
```

A spaced superscript also needs a space after it:

```bpl
2 3 ² + 1
```

### On a function

On a function, a superscript digit is a repeat count, as in [Power](glyphs/power.qmd). `f²` is `f⍣2`:

```bpl
inc←{⍵+1}
inc² 5
```

`f⁻¹` is the inverse of `f`, the same as `f⍣¯1`. `f⁻¹ Y` gives the `X` for which `f X` is `Y`. The inverse of the exponential `*` is the natural logarithm:

```bpl
*⁻¹1
```

Monadic Encode, `⊤`, gives a number's binary digits. Its inverse reads the digits back into a number:

```bpl
⊤⁻¹1 0 1
```

In a dyadic call, `X f⁻¹ Y` is the `Z` for which `X f Z` is `Y`. The inverses of `2×` and `2*` are division by 2 and the base-2 logarithm:

```bpl
2×⁻¹10
2*⁻¹8
```

Inverting [Replicate](glyphs/hash.qmd#expand) puts fill where the mask has zeros:

```bpl
1 0 1#⁻¹10 20
```

A superscript applies to the whole function on its left, as a monadic operator does. `+\⁻¹` turns running sums back into the numbers they came from:

```bpl
+\⁻¹1 3 6
```

The same goes for a derived function such as `-¨`:

```bpl
-¨⁻¹1 2 3
```

A bound function such as `2↣×` inverts as a whole too:

```bpl
2↣×⁻¹10
```

A superscript just after a number applies to the number, as a power. Without parentheses, `+↢3⁻¹` adds a third. To invert a function that ends in a number, put the function in parentheses:

```bpl
(+↢3)⁻¹ 10
```

A space before the superscript also works:

```bpl
+↢3 ⁻¹ 10
```

With the axis operator, the inverse can also go before `⍠`. `#⁻¹⍠¯1` is the same function as `(#⍠¯1)⁻¹`.

Inverses follow the rules of Power. They pass through bound arguments, Each, Rank and the other compositions that Power lists. [Inverse pair](glyphs/inverse-pair.qmd) attaches an inverse to any function.

### A function as a scalar

`fᵘ` is a scalar holding the function `f`, as [`⊂`](glyphs/enclose.qmd) encloses a subject. `⊂` can't enclose a function, because written next to one it forms a train. Like any scalar, `+ᵘ` has an empty shape:

```bpl
⍴+ᵘ
```

Its depth counts the one layer around the function:

```bpl
≡+ᵘ
```

### Errors

- `DOMAIN`: a function with no known inverse, `ᵀ` on a function, or `ᵘ` on an array
- `SYNTAX`: a superscript after a space that doesn't stand alone, as in `2 3 ²+1`

## Subscripts

A subscript selects a major cell of the item just before it. `v₁` is `1⌷v`:

```bpl
v₁
```

A negative subscript counts from the end:

```bpl
v₋₁
```

On a matrix, a subscript selects a row:

```bpl
m₁
```

After a dotted path, a subscript selects from the whole path:

```bpl
T←["a":[10 20]]
T.a₁
```

With a space before it, a subscript selects from the whole run on its left:

```bpl
2 3 ₁
```

As with a superscript, the run can be any expression:

```bpl
v+1 ₁
```

A spaced subscript also needs a space after it.

A subscripted item can be an assignment target:

```bpl
v₁←5
v
```

Modified assignment works too:

```bpl
v₁+←1
v
```

Use a subscript for one literal position on the leading axis. Use [Index](glyphs/squad.qmd), `⌷`, for a computed position, several positions, a key or another axis.

The subscript letters `ₓ`, `ⱼ`, `ₑ` and `ᵣ` are parts of numbers, as in `3ₓ` and `1ᵣ3`. [Numbers](numbers.qmd) covers them.

### Errors

- `SYNTAX`: a subscript after a space that doesn't stand alone, as in `2 3 ₁+1`, or that is an assignment target, as in `v ₁←5`
- `SYNTAX`: `₋` with no digit after it
- `INDEX`: the position is outside the leading axis
- `LENGTH`: the item is a unit, which has no axis to select from

## Differences from APL

BPL also accepts Dyalog's `x*2`, `f⍣2` and `f⍣¯1`. BQN writes the inverse as `𝔽⁼`. Neither language has subscripts.
