# Assignment

> Naming values, changing parts of arrays, and destructuring

`←` assigns the value to its right to the target, the run just before it. Besides plain names, a target can be a selection, a dot path, a subscript or a list of names. With a function just before `←`, as in `x+←1`, the assignment is modified. For the exact rules, see [Assign](glyphs/assign.qmd) and [Syntax](syntax.qmd#assignment).

## Plain assignment

`x←Y` binds `x` to `Y` and returns `Y` without displaying it:

```bpl
x←3
x+2
```

A value can be a function, such as `+/`:

```bpl
sum←+/
sum 1 2 3
```

Assignments chain:

```bpl
a←b←3
a+b
```

Put independent assignments on separate lines or between `⋄`, because `←` takes everything to its right:

```bpl
p←1 q←2
p
```

The name `_` discards a value. `_←Y` evaluates `Y` and keeps nothing.

## Modified assignment

`x f←Y` sets `x` to `x f Y`:

```bpl
x+←3
x
```

Any function works, such as `,` to append:

```bpl
s←"abc"
s,←"d"
s
```

A named function goes in parentheses, as `x(f)←Y`. In `x f←Y`, the target would be `f`:

```bpl
f←×
x(f)←4
x
```

## Assigning to part of an array

A [subscript](scripts.ipynb#subscripts) target replaces one literal position:

```bpl
v←1 2 3
v₁←9
v
```

Dot brackets name any positions, in plain or modified assignment:

```bpl
v.[[0 2]]+←10
v
```

Any other selection goes in parentheses, with a function that only selects or rearranges, as for structural [Under](glyphs/under.qmd):

```bpl
(2↑v)←8 9
v
```

The selection can apply a function with rank. `0⌷⍤1` selects the first item of each row:

```bpl
M←[1 2
   3 4]
(0⌷⍤1 M)←9
M
```

A dot path writes a field of a record:

```bpl
T←["n":1]
T.n+←1
T.n
```

A plain assignment to a dot path creates any missing records along the path:

```bpl
T.addr.city←"SF"
T
```

A subscript can follow a dot path:

```bpl
T.a←[1 2]
T.a₁←7
T.a
```

## Destructuring

A list of names in brackets destructures a value by position:

```bpl
[a b]←10 20
a+b
```

A single value goes to every name:

```bpl
[c d]←5
c d
```

A nested list of names destructures a nested value:

```bpl
[e [g h]]←[1 [2 3]]
h
```

Modified assignment works on a list of names too:

```bpl
[a b]+←10
a b
```

Inside brackets, every name is a target, even one that holds a function, such as `f`:

```bpl
[f n]←1 2
f
```

Against a vector with a key on every item, each name takes the item with its key:

```bpl
[b a]←["a":1 "b":2 "c":3]
a b
```

A program takes names from a [module](modules.qmd) this way, as in `[disp]←•load "lib/array.bpl"`.

## Names inside dfns

Inside a dfn, plain assignment makes a local name:

```bpl
total←0
set←{total←⍵ ⋄ total}
_←set 5
total
```

Modified and selective assignment update the nearest existing binding, even outside the dfn:

```bpl
add←{total+←⍵ ⋄ total}
_←add 5
total
```

Arrays have value semantics, as numbers do:

```bpl
u←1 2
w←u
u₀←9
w
```
