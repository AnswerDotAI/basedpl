# Regex
> Search and replace text with compiled regular expressions

```apl
]help •r
```

Patterns use [Rust regex syntax](https://docs.rs/regex/latest/regex/#syntax). Here `p` finds runs of digits:

```apl
p←•r "[0-9]+"
p.match "abc123def45"
p.position "abc123def45"
p.length "abc123def45"
'#' p.replace "abc123def45"
```

Search returns a vector, including zero or one match. Matches don't overlap. Lengths count characters:

```apl
p←•r "é|🐈+"
p.position "aé🐈🐈z"
p.length "aé🐈🐈z"
p.match "abc"
```

`groups` returns the captured strings of each match, without the whole match. An unmatched optional capture gives `""`:

```apl
p←•r "([a-z]+)([0-9]+)"
p.groups "ab12 cd3"
```

In a replacement, `$0` is the whole match and `$$` is a literal `$`. An unknown or unmatched capture expands to `""`:

```apl
p←•r "(?P<word>[a-z]+)([0-9]+)"
"${word}:$2:$$" p.replace "ab12"
```

`(?i)` ignores case, `(?m)` makes `^` and `$` match at line breaks, and `(?s)` lets `.` match newlines. Use Each to search several strings:

```apl
p←•r "(?i)cat"
(≢∘p.match)¨"Cat cat" "dog"
```

Empty matches occur at character boundaries:

```apl
p←•r ""
p.position "é🐈"
'-' p.replace "é🐈"
```

`lib/regex.apl` supplies `pattern (f regex_replace) text`. It applies `f` to each matched string and keeps the text between matches:

```apl
•load "lib/regex.apl"
"[0-9]+" ({⍕2×•json ⍵} regex_replace) "a12 b3"
```
